````python
import os, glob, time
import pandas as pd

def force_delete(path: str):
    # 1) снять "read-only" если вдруг выставлен
    try:
        os.chmod(path, 0o666)
    except Exception as e:
        print("chmod failed:", e)

    # 2) попытка обычного удаления
    try:
        os.remove(path)
        return True, "removed"
    except Exception as e:
        err1 = repr(e)

    # 3) если Google Drive иногда держит файл — переименуем и удалим
    try:
        tmp = path + f".to_delete_{int(time.time())}"
        os.rename(path, tmp)
        os.remove(tmp)
        return True, "renamed+removed"
    except Exception as e:
        err2 = repr(e)

    return False, f"remove failed: {err1}; rename+remove failed: {err2}"

def clean_checkpoints(output_dir: str, patterns=("*__paraphrase.csv", "*__backtranslate.csv")):
    candidates = []
    for pat in patterns:
        candidates += glob.glob(os.path.join(output_dir, pat))

    print("output_dir:", output_dir)
    print("found candidates:", len(candidates))

    removed = []
    for path in sorted(candidates):
        try:
            size = os.path.getsize(path)
        except FileNotFoundError:
            continue

        bad = False
        reason = None

        if size == 0:
            bad, reason = True, "size=0"
        else:
            try:
                df = pd.read_csv(path)
                if df.shape[1] == 0:
                    bad, reason = True, "no columns"
            except pd.errors.EmptyDataError:
                bad, reason = True, "EmptyDataError"
            except Exception as e:
                bad, reason = True, f"read_error:{type(e).__name__}"

        if bad:
            ok, msg = force_delete(path)
            print(f"[BAD] {path} ({reason}, size={size}) -> {msg}")
            if ok:
                removed.append(path)
        else:
            print(f"[OK ] {path} (size={size})")

    print("removed:", len(removed))
    return removed

# выполнить
_ = clean_checkpoints(output_dir)
````

````python
# ----------------- models
from sentence_transformers import SentenceTransformer
emb = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2", device=device)

from transformers import AutoTokenizer, AutoModelForCausalLM, AutoModelForSeq2SeqLM, BitsAndBytesConfig
import gc

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()

QWEN_ID = "Qwen/Qwen2.5-14B-Instruct"
qwen_tok = AutoTokenizer.from_pretrained(QWEN_ID, use_fast=True)

bnb = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_use_double_quant=True,
    bnb_4bit_compute_dtype=torch.float16,
    llm_int8_enable_fp32_cpu_offload=True,
)

max_memory = None
if device == "cuda":
    max_memory = {0: "14GiB", "cpu": "48GiB"}

qwen = AutoModelForCausalLM.from_pretrained(
    QWEN_ID,
    device_map="auto" if device == "cuda" else None,
    max_memory=max_memory,
    quantization_config=bnb if device == "cuda" else None,
    torch_dtype=torch.float16 if device == "cuda" else None,
    offload_folder="./offload_qwen" if device == "cuda" else None,
)
qwen.eval()

NLLB_ID = "facebook/nllb-200-distilled-600M"
nllb_tok = AutoTokenizer.from_pretrained(NLLB_ID)
nllb = AutoModelForSeq2SeqLM.from_pretrained(
    NLLB_ID,
    device_map="auto" if device == "cuda" else None,
    torch_dtype=torch.float16 if device == "cuda" else None,
)
nllb.eval()

# ----------------- helpers
# плейсхолдеры вида [ORGANIZATION]
PLACEHOLDER_RE = re.compile(r"$$[A-Z_]+$$")   # FIX: было $$...$$

def placeholders_signature(text: str):
    return tuple(sorted(PLACEHOLDER_RE.findall(str(text))))

def safe_name(s: str, max_len: int = 120) -> str:
    s = re.sub(r"[^\w\-]+", "_", s, flags=re.UNICODE)
    s = re.sub(r"_+", "_", s).strip("_")
    return s[:max_len] if len(s) > max_len else s

def cosine_dedup(new_texts, existing_texts, sim_th=0.92):
    if not new_texts:
        return []
    exist_vec = emb.encode(existing_texts, convert_to_numpy=True, normalize_embeddings=True)
    kept_vec = exist_vec
    kept = []
    for t in new_texts:
        v = emb.encode([t], convert_to_numpy=True, normalize_embeddings=True)[0]
        sims = kept_vec @ v
        if sims.max() < sim_th:
            kept.append(t)
            kept_vec = np.vstack([kept_vec, v])
    return kept

def _cuda_cleanup():
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

@torch.no_grad()
def qwen_paraphrase_one(text: str, max_new_tokens=220):
    """
    Возвращает строку или None, если словили OOM (чтобы не терять прогресс).
    Также формирует attention_mask, чтобы не было warning и странного поведения.
    """
    user_prompt = (
        "Перефразируй текст письма на русском.\n"
        "Требования:\n"
        "- Сохрани смысл и категорию письма\n"
        "- Не добавляй новых фактов\n"
        "- Токены в квадратных скобках вида [NAME], [DOCUMENT_NUMBER], [DATE_TIME] не изменяй и не переставляй\n"
        "- Ответ дай только перефразированным текстом, без комментариев\n\n"
        f"Текст:\n{text}\n\nПерефраз:"
    )
    messages = [
        {"role": "system", "content": "Ты аккуратный помощник по перефразированию официальных писем на русском."},
        {"role": "user", "content": user_prompt},
    ]

    # FIX: делаем токенизацию через tokenizer(...), чтобы получить attention_mask
    rendered = qwen_tok.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    batch = qwen_tok(
        rendered,
        return_tensors="pt",
        truncation=True,
        max_length=1024,
        padding=False,
    )

    if device == "cuda":
        batch = {k: v.to(qwen.device) for k, v in batch.items()}
    else:
        batch = {k: v.to(device) for k, v in batch.items()}

    try:
        out = qwen.generate(
            **batch,
            max_new_tokens=max_new_tokens,
            do_sample=False,
            num_beams=1,
            pad_token_id=qwen_tok.eos_token_id,
        )
        return qwen_tok.decode(out[0][batch["input_ids"].shape[-1]:], skip_special_tokens=True).strip()

    except torch.cuda.OutOfMemoryError:
        _cuda_cleanup()
        return None

@torch.no_grad()
def nllb_translate(texts, src_lang: str, tgt_lang: str, max_new_tokens=256, batch_size=8):
    out_texts = []
    nllb_tok.src_lang = src_lang
    forced_bos = nllb_tok.convert_tokens_to_ids(tgt_lang)

    for i in range(0, len(texts), batch_size):
        batch = texts[i:i+batch_size]
        inputs = nllb_tok(batch, return_tensors="pt", padding=True, truncation=True, max_length=512)
        if device == "cuda":
            inputs = {k: v.to(nllb.device) for k, v in inputs.items()}
        else:
            inputs = {k: v.to(device) for k, v in inputs.items()}

        gen = nllb.generate(
            **inputs,
            forced_bos_token_id=forced_bos,
            max_new_tokens=max_new_tokens,
            num_beams=4,
            do_sample=False,
        )
        out_texts.extend(nllb_tok.batch_decode(gen, skip_special_tokens=True))
    return out_texts

def back_translate_ru_en_ru(texts):
    en = nllb_translate(texts, src_lang="rus_Cyrl", tgt_lang="eng_Latn")
    ru = nllb_translate(en, src_lang="eng_Latn", tgt_lang="rus_Cyrl")
    return ru

def load_phase_csv(path: str):
    if os.path.exists(path):
        try:
            dfp = pd.read_csv(path)
        except pd.errors.EmptyDataError:
            return pd.DataFrame(columns=["text", "label", "src"])

        if "text" not in dfp.columns:
            dfp["text"] = []
        if "label" not in dfp.columns:
            dfp["label"] = []
        if "src" not in dfp.columns:
            dfp["src"] = []
        return dfp[["text", "label", "src"]]
    return None

def save_phase_csv(df_phase: pd.DataFrame, path: str):
    if df_phase is None or len(df_phase) == 0:
        df_phase = pd.DataFrame(columns=["text", "label", "src"])
    df_phase.to_csv(path, index=False)

# ----------------- load train
df = pd.read_csv(train_path)
df["text"] = df["text"].astype(str)
df["label"] = df["label"].astype(str)

vc = df["label"].value_counts()
small_labels = set(vc[vc < 50].index)
print("train:", df.shape, "labels:", df["label"].nunique(), "classes<50:", len(small_labels))

# ----------------- per-class processing with checkpoints
all_rows = []

for label, cnt in vc.items():
    df_lbl = df[df["label"] == label].copy()
    orig_texts = df_lbl["text"].tolist()
    orig_sigs = set(placeholders_signature(t) for t in orig_texts)

    for t in orig_texts:
        all_rows.append({"text": t, "label": label, "src": "orig"})

    if label not in small_labels:
        continue

    cls_name = safe_name(label)
    paraphrase_path = os.path.join(output_dir, f"{cls_name}__paraphrase.csv")
    backtr_path     = os.path.join(output_dir, f"{cls_name}__backtranslate.csv")

    # ---------- PHASE 1: paraphrase to 30
    phase1 = load_phase_csv(paraphrase_path)
    if phase1 is None:
        need = max(0, 30 - cnt)
        gen = []
        t0 = time.time()

        # FIX: устойчиво к OOM + периодическая чистка VRAM
        tries = 0
        i = 0
        while len(gen) < need and tries < need * 3:
            src = orig_texts[i % len(orig_texts)]
            g = qwen_paraphrase_one(src)
            tries += 1
            i += 1

            if g is None:
                continue

            gen.append(g)

            if device == "cuda" and (len(gen) % 3 == 0):
                _cuda_cleanup()

        print(f"[{label}] paraphrase generated={len(gen)} tries={tries} time={time.time()-t0:.1f}s")

        gen = [t for t in gen if placeholders_signature(t) in orig_sigs]
        gen = cosine_dedup(gen, orig_texts, sim_th=0.92)

        phase1 = pd.DataFrame([{"text": t, "label": label, "src": "paraphrase"} for t in gen],
                              columns=["text", "label", "src"])
        save_phase_csv(phase1, paraphrase_path)
        print(f"[{label}] paraphrase kept={len(phase1)} saved={paraphrase_path}")
    else:
        print(f"[{label}] paraphrase SKIP (exists) {paraphrase_path}")

    para_texts = phase1["text"].astype(str).tolist() if "text" in phase1.columns else []
    for t in para_texts:
        all_rows.append({"text": t, "label": label, "src": "paraphrase"})

    # ---------- PHASE 2: back-translate to 50
    cur = cnt + len(para_texts)

    phase2 = load_phase_csv(backtr_path)
    if phase2 is None:
        need = max(0, 50 - cur)
        bt_inputs = [orig_texts[i % len(orig_texts)] for i in range(need)]
        t0 = time.time()
        gen_bt = back_translate_ru_en_ru(bt_inputs)
        print(f"[{label}] backtranslate generated={len(gen_bt)} time={time.time()-t0:.1f}s")

        gen_bt = [t for t in gen_bt if placeholders_signature(t) in orig_sigs]
        base_for_dedup = orig_texts + para_texts
        gen_bt = cosine_dedup(gen_bt, base_for_dedup, sim_th=0.92)

        phase2 = pd.DataFrame([{"text": t, "label": label, "src": "backtranslate"} for t in gen_bt],
                              columns=["text", "label", "src"])
        save_phase_csv(phase2, backtr_path)
        print(f"[{label}] backtranslate kept={len(phase2)} saved={backtr_path}")
    else:
        print(f"[{label}] backtranslate SKIP (exists) {backtr_path}")

    bt_texts = phase2["text"].astype(str).tolist() if "text" in phase2.columns else []
    for t in bt_texts:
        all_rows.append({"text": t, "label": label, "src": "backtranslate"})
````

````python
import os, glob
import pandas as pd
import numpy as np

def build_final_from_checkpoints(
    train_path: str,
    output_dir: str,
    final_out_path: str,
    small_class_threshold: int = 50,
    within_class_sim_th: float = 0.95,
    report_top_n: int = 50,
):
    # 1) originals
    df = pd.read_csv(train_path)
    df["text"] = df["text"].astype(str)
    df["label"] = df["label"].astype(str)

    orig = df[["text", "label"]].copy()
    orig["src"] = "orig"

    # 2) checkpoints
    ckpt_paths = sorted(
        glob.glob(os.path.join(output_dir, "*__paraphrase.csv")) +
        glob.glob(os.path.join(output_dir, "*__backtranslate.csv"))
    )

    parts = [orig]
    bad = []

    for p in ckpt_paths:
        try:
            d = pd.read_csv(p)
        except pd.errors.EmptyDataError:
            bad.append((p, "EmptyDataError"))
            continue
        except Exception as e:
            bad.append((p, type(e).__name__))
            continue

        if d is None or len(d) == 0:
            continue

        if "text" not in d.columns:
            bad.append((p, "no_text_col"))
            continue
        if "label" not in d.columns:
            bad.append((p, "no_label_col"))
            continue
        if "src" not in d.columns:
            d["src"] = "paraphrase" if p.endswith("__paraphrase.csv") else "backtranslate"

        d = d[["text", "label", "src"]].copy()
        d["text"] = d["text"].astype(str)
        d["label"] = d["label"].astype(str)
        parts.append(d)

    if bad:
        print("WARNING: some checkpoints were skipped:")
        for p, r in bad[:20]:
            print(" -", p, "=>", r)
        if len(bad) > 20:
            print(f" ... and {len(bad)-20} more")

    aug_df = pd.concat(parts, ignore_index=True)

    # 3) exact dedup
    before = len(aug_df)
    aug_df = aug_df.drop_duplicates(subset=["label", "text"]).reset_index(drop=True)
    print("exact dedup:", before, "->", len(aug_df))

    # 4) cosine-dedup within small classes (uses global emb)
    vc_train = df["label"].value_counts()
    small_labels = set(vc_train[vc_train < small_class_threshold].index)

    def cosine_dedup_within_class(df_class, sim_th=0.95):
        texts = df_class["text"].astype(str).tolist()
        if len(texts) <= 1:
            return df_class
        vecs = emb.encode(texts, convert_to_numpy=True, normalize_embeddings=True)
        keep = []
        kept_vecs = []
        for i, v in enumerate(vecs):
            if not kept_vecs:
                keep.append(i); kept_vecs.append(v); continue
            sims = np.dot(np.vstack(kept_vecs), v)
            if sims.max() < sim_th:
                keep.append(i); kept_vecs.append(v)
        return df_class.iloc[keep]

    out_parts = []
    for label, part in aug_df.groupby("label"):
        if label in small_labels:
            out_parts.append(cosine_dedup_within_class(part, sim_th=within_class_sim_th))
        else:
            out_parts.append(part)

    aug_df = pd.concat(out_parts, ignore_index=True)

    # 5) coverage check: each label should have >= threshold
    vc_final = aug_df["label"].value_counts()
    missing = vc_final[vc_final < small_class_threshold].sort_values()

    if len(missing) > 0:
        print("\nWARNING: some labels still have < "
              f"{small_class_threshold} samples after augmentation:")
        print(missing.to_string())

        # helpful: show which checkpoint files correspond to those labels (by safe_name)
        def safe_name(s: str, max_len: int = 120) -> str:
            s = re.sub(r"[^\w\-]+", "_", s, flags=re.UNICODE)
            s = re.sub(r"_+", "_", s).strip("_")
            return s[:max_len] if len(s) > max_len else s

        print("\nSuggested checkpoint files to delete for re-run (if you want to regenerate):")
        for label in missing.index[:report_top_n]:
            cls = safe_name(label)
            p1 = os.path.join(output_dir, f"{cls}__paraphrase.csv")
            p2 = os.path.join(output_dir, f"{cls}__backtranslate.csv")
            print(f"- {label} ({int(vc_final[label])}):")
            print(f"  {p1} {'[exists]' if os.path.exists(p1) else '[missing]'}")
            print(f"  {p2} {'[exists]' if os.path.exists(p2) else '[missing]'}")
        if len(missing) > report_top_n:
            print(f"... and {len(missing)-report_top_n} more")

        # не сохраняем "как будто всё ок" молча — но файл всё равно пишем, чтобы можно было посмотреть
    else:
        print(f"\nOK: all labels have >= {small_class_threshold} samples.")

    # 6) save
    aug_df.to_csv(final_out_path, index=False)
    print("saved:", final_out_path, "shape:", aug_df.shape)

    return aug_df, missing

build_final_from_checkpoints(train_path, output_dir, final_out_path)
````

Код делает аугментацию датасета писем: для “малых” классов (где мало примеров) он генерирует дополнительные тексты двумя способами (перефразирование Qwen и back-translation через NLLB), фильтрует их по плейсхолдерам и по семантической похожести, сохраняет промежуточные результаты в чекпойнты на Google Drive и в конце собирает общий CSV.

Ниже — что происходит **по блокам**.

---

## 1) Models: загрузка моделей
### 1.1 SentenceTransformer (эмбеддинги)
```python
emb = SentenceTransformer("all-MiniLM-L6-v2", device=device)
```
Нужен для вычисления эмбеддингов текста и **косинусного дедупа** (убрать слишком похожие генерации).

### 1.2 Qwen2.5-14B-Instruct (перефразирование)
- Загружается токенизатор `qwen_tok`
- Загружается модель `qwen` в 4-bit (через `BitsAndBytesConfig`)
- `device_map="auto"` распределяет модель по GPU/CPU при необходимости
- `max_memory` ограничивает, сколько памяти можно занять на GPU/CPU
- `offload_folder` — папка для выгрузки частей модели на диск (если понадобится)

Цель: генерировать **перефраз** исходного письма.

### 1.3 NLLB-200 distilled 600M (перевод)
Загружается `nllb_tok` и `nllb` для машинного перевода.
Цель: делать **обратный перевод** RU→EN→RU как способ аугментации.

---

## 2) Helpers: утилиты и фильтры

### 2.1 Поиск плейсхолдеров
```python
PLACEHOLDER_RE = r"\[[A-Z_]+\]"
```
Ищет токены вида `[ORGANIZATION]`, `[ID]`, `[DATE_TIME]`.

`placeholders_signature(text)` возвращает “сигнатуру” — отсортированный список всех таких токенов в тексте.  
Это нужно, чтобы **не допускать генерации, где модель испортила/удалила/добавила плейсхолдеры**.

### 2.2 safe_name
Делает безопасное имя файла из названия класса (заменяет пробелы/кавычки на `_`), чтобы сохранять чекпойнты.

### 2.3 cosine_dedup
Берёт:
- `new_texts` (сгенерированные)
- `existing_texts` (база: оригиналы или оригиналы+перефразы)

Считает эмбеддинги и оставляет только те новые тексты, которые:
- не слишком похожи на базу
- и не слишком похожи друг на друга  
Порог `sim_th=0.92` — чем выше, тем “строже” удаление похожих.

---

## 3) Генерация перефраза Qwen (qwen_paraphrase_one)
Функция:
1) Формирует инструкцию (что делать и что нельзя делать с плейсхолдерами).
2) Оборачивает в chat-template (system+user).
3) Токенизирует так, чтобы получить `attention_mask` (это важно для стабильности).
4) Вызывает `qwen.generate(...)`.
5) Декодирует **только сгенерированную часть** (после входных токенов).

### Защита от OOM
Если случился `torch.cuda.OutOfMemoryError`, функция:
- чистит память (`gc.collect()` + `torch.cuda.empty_cache()`)
- возвращает `None`, чтобы основной цикл **не падал**, а просто пропускал неудачную попытку.

---

## 4) Перевод NLLB и back-translation
### 4.1 nllb_translate
Переводит список текстов батчами:
- задаёт `src_lang`
- задаёт `forced_bos_token_id` для языка назначения
- генерирует перевод `nllb.generate(...)`

### 4.2 back_translate_ru_en_ru
Делает:
1) RU (`rus_Cyrl`) → EN (`eng_Latn`)
2) EN → RU  
Это даёт “перефраз” через перевод.

---

## 5) Checkpoints: чтение/запись промежуточных CSV
### load_phase_csv(path)
- Если файла нет → `None` (значит фазу надо генерировать)
- Если файл есть, но пустой → возвращает пустой DataFrame с колонками `text,label,src`
- Если файл нормальный → читает и гарантирует наличие колонок

### save_phase_csv(df, path)
Сохраняет DataFrame в CSV. Даже если он пустой — пишет корректный CSV с заголовками.

Смысл чекпойнтов: вы можете **перезапускать Colab много раз**, и уже сделанные классы будут пропускаться.

---

## 6) Load train: загрузка исходного датасета
Определяет “малые” классы, которые нужно доаугментировать до 50 примеров.

---

## 7) Основной цикл по классам
Для каждого `label`:

### 7.1 Добавляет оригиналы в итоговый список
Оригинальные тексты всегда попадают в `all_rows` с `src="orig"`.

### 7.2 Если класс не малый — пропускает
Если `label` не в `small_labels`, генерации не делаются.

### 7.3 PHASE 1: перефразирование до 30
- Если чекпойнт `__paraphrase.csv` уже есть → SKIP
- Иначе:
  - `need = max(0, 30 - cnt)`
  - пытается сгенерировать `need` перефразов
  - если OOM → попытка не засчитывается, пробует дальше (но ограничено `tries < need*3`)
  - фильтрует по плейсхолдерам (сигнатура должна совпасть с одной из сигнатур оригиналов)
  - cosine_dedup против оригиналов
  - сохраняет чекпойнт

### 7.4 PHASE 2: back-translation до 50
- Считает текущий размер класса: `cur = cnt + len(para_texts)`
- Если чекпойнт `__backtranslate.csv` есть → SKIP
- Иначе:
  - `need = max(0, 50 - cur)`
  - делает back-translation для `need` текстов
  - фильтр по плейсхолдерам
  - cosine_dedup против (оригиналы + перефразы)
  - сохраняет чекпойнт

---

## 8) Финальная сборка и дедуп
### 8.1 Точный дедуп
Удаляет точные дубликаты по паре `(label, text)`.

### 8.2 Доп. cosine-дедуп внутри малых классов
Для каждого малого класса ещё раз удаляет слишком похожие тексты внутри класса (`sim_th=0.95`).

### 8.3 Сохранение финального датасета
Пишет `aug_df` в `final_out_path`.
